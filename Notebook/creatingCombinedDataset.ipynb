```python
import pandas as pd
```

```python
npl_2024 = pd.read_csv("./Datasets/NPL-2024.csv")
npl_2025 = pd.read_csv("./Datasets/NPL-2025.csv")
```

```python
npl_2024.columns.tolist()
```

```python
npl_2025.columns.tolist()
```

```python
npl_2025.rename(columns={
    "match_no":"match_id",
    "batter":"batsman",
    "player_out":"player_dismissed",
    "fielders":"fielder",
    "match_no":"match_id",
    "match_no":"match_id",
} , inplace=True)
npl_2025.columns.tolist()
```

```python
npl_2024["match_id"] = "2024_"+npl_2024["match_id"].astype(str)
npl_2025["match_id"] = "2025_"+npl_2025["match_id"].astype(str)
```

```python
npl_dataset = pd.concat([npl_2024 , npl_2025] , ignore_index=True)
```

```python
npl_dataset.to_csv("./Datasets/npl_dataset.csv")
```
