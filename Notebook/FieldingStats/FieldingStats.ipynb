```python
import pandas as pd
```

```python
df = pd.read_csv("../Datasets/npl_dataset.csv")
df.sample(5)
```

```python
#Most Catches Taken

catch_df = df[df["dismissal_kind"] == "caught"]
total_catch = catch_df.groupby("fielder").size().reset_index(name="Catches").sort_values(by="Catches" , ascending=False)
total_catch.head(10)
```

```python
#Most Run Outs

runout_df = df[df["dismissal_kind"] == "run out"]
total_runouts = runout_df.groupby("fielder").size().reset_index(name="Runouts").sort_values(by="Runouts" , ascending=False)
total_runouts.head(10)
```

```python
#Most Stumped
stumpped_df = df[df["dismissal_kind"] == "stumped"]
total_stumpped = stumpped_df.groupby("fielder").size().reset_index(name="Stumpped").sort_values(by="Stumpped" , ascending=False)
total_stumpped.head(10)
```
